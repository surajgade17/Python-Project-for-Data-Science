```python
import yfinance as yf
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

# Question 1: Extracting Tesla Stock Data Using yfinance

In this section, we use the yfinance library to extract historical stock data for Tesla.

Tesla's stock ticker symbol is TSLA.

We retrieve the maximum available historical data and reset the DataFrame index so that the Date becomes a regular column.


```python
tesla = yf.Ticker("TSLA")

tesla_data = tesla.history(period="max")

tesla_data.reset_index(inplace=True)

tesla_data.head()
```

### Tesla Stock Data

The table above displays the first five rows of Tesla's historical stock data.

Important columns include:
- Date
- Open
- High
- Low
- Close
- Volume


```python
print("Tesla stock data shape:", tesla_data.shape)
print("\nTesla stock data columns:")
print(tesla_data.columns.tolist())
```

# Question 2: Extracting Tesla Revenue Data Using Web Scraping

In this section, we use the requests library and BeautifulSoup to extract Tesla's quarterly revenue data from an HTML table.

The revenue values are then cleaned by removing dollar signs and commas and converting them into numeric values.


```python
tesla_revenue_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0220EN-SkillsNetwork/labs/project/revenue.htm"

response = requests.get(tesla_revenue_url)

print("Status code:", response.status_code)
```

```python
tesla_revenue = pd.read_html(tesla_revenue_url, match="Tesla Quarterly Revenue")[0]
tesla_revenue.columns = ["Date", "Revenue"]

print("Number of rows scraped:", len(tesla_revenue))
tesla_revenue.head()
```

```python
# (Data already loaded into tesla_revenue above using pd.read_html)
tesla_revenue.tail()
```

### Cleaning Tesla Revenue Data

The revenue column may contain dollar signs, commas, and empty values. We remove these characters and convert the column to numeric format.


```python
tesla_revenue["Revenue"] = (
    tesla_revenue["Revenue"]
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
)

tesla_revenue["Revenue"] = pd.to_numeric(
    tesla_revenue["Revenue"],
    errors="coerce"
)

tesla_revenue.dropna(inplace=True)
tesla_revenue.tail()
```

```python
print("Tesla revenue data shape:", tesla_revenue.shape)
```

# Question 4: Extracting GameStop Revenue Data Using Web Scraping

In this section, we scrape GameStop's quarterly revenue data from an HTML table.

We then clean the revenue column by removing dollar signs and commas and convert the values into numeric data.


```python
gme_revenue_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0220EN-SkillsNetwork/labs/project/stock.html"

response = requests.get(gme_revenue_url)

print("Status code:", response.status_code)
```

```python
gme_revenue = pd.read_html(gme_revenue_url, match="GameStop Quarterly Revenue")[0]
gme_revenue.columns = ["Date", "Revenue"]

print("Number of rows scraped:", len(gme_revenue))
gme_revenue.head()
```

```python
gme_revenue["Revenue"] = (
    gme_revenue["Revenue"]
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
)

gme_revenue["Revenue"] = pd.to_numeric(
    gme_revenue["Revenue"],
    errors="coerce"
)

gme_revenue.dropna(inplace=True)
gme_revenue.tail()
```

```python
print("GameStop revenue data shape:", gme_revenue.shape)
print("\nLast five rows:")
gme_revenue.tail()
```

# Question 5: Tesla Stock and Revenue Dashboard

The following dashboard contains two graphs:

1. Tesla historical share price.
2. Tesla quarterly revenue.

The dashboard allows us to visually compare Tesla's stock performance with its reported revenue over time.


```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots
```

```python
fig = make_subplots(
    rows=2,
    cols=1,
    shared_xaxes=False,
    subplot_titles=(
        "Tesla Historical Stock Price",
        "Tesla Quarterly Revenue"
    ),
    vertical_spacing=0.15
)

fig.add_trace(
    go.Scatter(
        x=tesla_data["Date"],
        y=tesla_data["Close"],
        mode="lines",
        name="Tesla Stock Price",
        line=dict(color="blue")
    ),
    row=1,
    col=1
)

fig.add_trace(
    go.Scatter(
        x=tesla_revenue["Date"],
        y=tesla_revenue["Revenue"],
        mode="lines+markers",
        name="Tesla Revenue",
        line=dict(color="green")
    ),
    row=2,
    col=1
)

fig.update_layout(
    title="Tesla Stock Price and Revenue Dashboard",
    height=800,
    template="plotly_white",
    showlegend=True
)

fig.update_yaxes(
    title_text="Stock Price (USD)",
    row=1,
    col=1
)

fig.update_yaxes(
    title_text="Revenue (USD)",
    row=2,
    col=1
)

fig.update_xaxes(
    title_text="Date",
    row=2,
    col=1
)

fig.show()
```

### Tesla Dashboard Result

The first graph shows Tesla's historical closing stock price.

The second graph shows Tesla's quarterly revenue.

The dashboard provides a visual comparison between Tesla's market performance and revenue over time.
